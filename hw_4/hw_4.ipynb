
### Задание 1. Проверка автомобильного номера (синтаксис)

Напишите функцию `check_car_id(car_id: str) -> str`, которая проверяет корректность автомобильного номера РФ.

**Формат номера:**

- 1 буква + 3 цифры + 2 буквы + 2–3 цифры региона.
- Допустимые буквы: `А, В, Е, К, М, Н, О, Р, С, Т, У, Х` (кириллица).
- Учитывайте путаницу между кириллицей и латиницей. Выберите один подход: нормализовать латинские буквы или считать их невалидными.

**Результат:**

- Валидный номер: `Номер А222ВС валиден. Регион: 96.`
- Невалидный номер: `Номер не валиден.`

#### Пример работы программы

**Input:**

```python
check_car_id('А222ВС96')
check_car_id('А123ВЕ777')
check_car_id('АБ22ВВ193')
```

**Output:**

```text
Номер А222ВС валиден. Регион: 96.
Номер А123ВЕ валиден. Регион: 777.
Номер не валиден.
```

#### Пограничные случаи

**Input:**

```python
check_car_id('A222BC96')    # Латиница A/B/C
check_car_id('А222ВС9')     # Регион из 1 цифры
check_car_id('А222ВС9699')  # Регион из 4 цифр
check_car_id('Я222ВС96')    # Недопустимая буква
check_car_id('')            # Пустая строка
```

**Output:**

```text
Номер А222ВС валиден. Регион: 96.
Номер не валиден.
Номер не валиден.
Номер не валиден.
Номер не валиден.
```

*В примере используется нормализация латиницы. При строгой проверке кириллицы первый результат должен быть «Номер не валиден.».*


### Решение

````python
import re

values_ru = 'АВЕКМНОРСТУХ'
values_lat = 'ABEKMHOPCTYX'

def check_car_id(car_id):
    # Convert Latin letters to Cyrillic
    translation = str.maketrans(values_lat, values_ru)
    car_id = car_id.upper().translate(translation)

    # Check the full number format
    pattern = rf'([{values_ru}])(\d{{3}})([{values_ru}]{{2}})(\d{{2,3}})'
    res = re.fullmatch(pattern, car_id)

    if res:
        number = ''.join(res.groups()[:3])
        region = res.group(4)
        return f'Номер {number} валиден. Регион: {region}.'

    return 'Номер не валиден.'
````

````python

print(check_car_id('А222ВС96'))
print(check_car_id('А123ВЕ777'))
print(check_car_id('АБ22ВВ193'))
print(check_car_id('JK22ВВ193'))
print(check_car_id('А222ВС9'))
print(check_car_id('А222ВС9699'))
print(check_car_id('Я222ВС96'))
print(check_car_id(''))
````
