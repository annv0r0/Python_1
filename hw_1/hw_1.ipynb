 ```python

 ```

### Решение 1

````python
from pathlib import Path
import json

purchase_path = Path('purchase_log.txt')
purchases = {}
````

````python
with open(purchase_path) as f:
    for i, line in enumerate(f):
        if i > 0:
            content = json.loads(line)
            purchases[content['user_id']] = content['category']
````

### Решение 2

````python
from pathlib import Path
import csv
import json

visit_path = Path('visit_log.csv')
purchase_path = Path('purchase_log.txt')
funnel_path = Path('funnel.csv')

purchases = {}
````

````python
with open(purchase_path) as f:
    for i, line in enumerate(f):
        if i > 0:
            content = json.loads(line)
            purchases[content['user_id']] = content['category']
````

````python
with (
    open(visit_path) as visit_file,
    open(funnel_path, "w", newline="\n") as funnel_file
):
    reader = csv.reader(visit_file)
    writer = csv.writer(funnel_file)
    
    for i, visit in enumerate(reader):
        if i == 0: 
            # header
            writer.writerow(["user_id", "source", "category"])
        else:
            uid = visit[0]
            if uid in purchases:
                res = [uid, visit[1], purchases[uid]]
                writer.writerow([uid, visit[1], purchases[uid]])

# find funnel.csv in the current dir
````
